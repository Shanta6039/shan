## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
# Plan (in plain words):
# - Store spending as a dictionary keyed by category.
# - Each category maps to a small dictionary holding a running 'total'
#   and a list of the individual 'transactions' in that category.
# - This means total_for() is a simple lookup, summary() is a simple
#   extraction of totals, and add_transaction() just updates one entry
#   (creating it the first time a category is seen).

spending = {}

def add_transaction(spending, transaction):
    """Add one transaction dict to the spending structure, updating its
    category's running total and transaction list. Creates the category
    entry the first time it is seen."""
    category = transaction['category']
    amount = transaction['amount']
    if category not in spending:
        spending[category] = {'total': 0, 'transactions': []}
    spending[category]['total'] += amount
    spending[category]['transactions'].append(transaction)

def total_for(spending, category):
    """Return the total spent in a category, or 0 if it was never recorded."""
    return spending.get(category, {}).get('total', 0)

def summary(spending):
    """Return a plain dict mapping every category to its total."""
    return {category: data['total'] for category, data in spending.items()}

# Build the dictionary from the raw transaction list.
for t in transactions:
    add_transaction(spending, t)

summary(spending)
```

```python
# My feature: which category do I spend the most on, and which single
# transactions are "big" (above a threshold I choose)?

def top_category(spending):
    """Return the (category, total) pair with the highest total spend.
    Returns (None, 0) if spending is empty."""
    if not spending:
        return (None, 0)
    best_category = max(spending, key=lambda c: spending[c]['total'])
    return (best_category, spending[best_category]['total'])

def transactions_above(spending, threshold):
    """Return a list of every individual transaction (across all
    categories) whose amount is strictly greater than threshold."""
    big = []
    for data in spending.values():
        for t in data['transactions']:
            if t['amount'] > threshold:
                big.append(t)
    return big

print("Biggest spending category:", top_category(spending))
print("Transactions over $50:")
for t in transactions_above(spending, 50):
    print(" ", t['date'], t['category'], t['amount'], '-', t['description'])
```

### Your reflection

**Why did you structure your dictionary the way you did?**
I keyed the dictionary by category, since that is the thing the three required functions (total_for, add_transaction, summary) all care about first. Each category's value is itself a small dictionary holding a running total plus the list of raw transactions in that category. Keeping the raw transactions around (not just the total) is what let me build the "biggest category" and "transactions above a threshold" features without re-deriving anything.

**What kind of question would your structure make awkward to answer?** Anything organised by time rather than category, e.g. "how much did I spend in the third week of August?" I'd have to loop over every category and every transaction and check the date each time, since dates aren't indexed anywhere. If that were a common question I'd add a second dictionary keyed by week or date instead.

**If a transaction came through with no category on it, what would your lookup do?** transaction['category'] would raise a KeyError in add_transaction before it even got stored, since I access the key directly rather than with .get(). If I used transaction.get('category') instead it would silently file the transaction under a None key, which is arguably worse because it hides the problem. total_for itself never crashes on an unknown category, missing or otherwise, because it uses .get(..., {}) with a default.
