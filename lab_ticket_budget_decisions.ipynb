## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Your task is to create a Python program that acts as an 'Expense Classifier and Budget Checker'. This tool will help a user understand their spending habits and warn them if an expense is too large a share of their monthly budget. The program should be interactive, clearly ask for user input, process the data, and provide specific feedback.

Here are the detailed requirements:

1.  **Welcome Message**: Start with a friendly welcome message, explaining the purpose of the tool.

2.  **User Input**: The program must ask the user for the following information, one by one:
    *   `monthly_income`: An integer representing their total monthly income.
    *   `monthly_budget_percentage`: An integer (between 1 and 100) representing the percentage of their income they want to allocate as their total monthly budget for expenses.
    *   `expense_amount`: A floating-point number representing the current expense they are considering.
    *   `expense_category`: A string representing the category of the expense (e.g., 'Groceries', 'Utilities', 'Entertainment', 'Transport', 'Rent', 'Savings', 'Other').

3.  **Input Validation**: Ensure all numerical inputs are valid (e.g., positive numbers, percentage between 1-100). If input is invalid, prompt the user again until valid input is received.

4.  **Calculations**:
    *   `total_monthly_budget`: Calculate this as `monthly_income * (monthly_budget_percentage / 100)`.
    *   `remaining_budget`: For this single expense, calculate `total_monthly_budget - expense_amount`.

5.  **Expense Classification and Feedback**: Based on the `expense_category`, provide specific feedback:
    *   **Groceries**: Always a necessary expense. Suggest looking for deals if `expense_amount` is greater than 20% of `total_monthly_budget`.
    *   **Utilities**: Necessary. Warn if `expense_amount` is greater than 15% of `total_monthly_budget`.
    *   **Entertainment**: Discretionary. Advise caution if `expense_amount` is greater than 10% of `total_monthly_budget`. Suggest alternative free activities if `expense_amount` is greater than 20% of `total_monthly_budget`.
    *   **Transport**: Essential for most. Warn if `expense_amount` is greater than 15% of `total_monthly_budget`.
    *   **Rent**: Fixed and often large. Simply state that it's a significant fixed cost. If `expense_amount` is greater than 30% of `monthly_income`, issue a strong warning about affordability.
    *   **Savings**: Always positive. Congratulate the user on saving.
    *   **Other**: Generic advice. If `expense_amount` is greater than 10% of `total_monthly_budget`, suggest reviewing this expense carefully.

6.  **Overall Budget Check**: After category-specific feedback, provide general feedback based on the `expense_amount` relative to the `total_monthly_budget`:
    *   If `expense_amount` is less than or equal to `total_monthly_budget`:
        *   If `expense_amount` is less than 5% of `total_monthly_budget`: "This expense is a very small part of your budget. Good job!"
        *   If `expense_amount` is between 5% and 20% of `total_monthly_budget`: "This expense fits comfortably within your budget."
        *   If `expense_amount` is greater than 20% of `total_monthly_budget`: "This expense takes a significant portion of your budget. Consider if it's truly necessary."
    *   If `expense_amount` is greater than `total_monthly_budget`: "Warning! This expense exceeds your monthly budget. You need to adjust your spending or reconsider this purchase."

7.  **Summary**: Display the `monthly_income`, `total_monthly_budget`, `expense_amount`, `expense_category`, and `remaining_budget` at the end.
```

````python
# Paste the code your prompt produced here, then run it to show it works.

def get_valid_input(prompt, type_func, validation_func=None, error_message="Invalid input. Please try again."):
    """Helper function to get and validate user input."""
    while True:
        try:
            value = type_func(input(prompt))
            if validation_func and not validation_func(value):
                print(error_message)
            else:
                return value
        except ValueError:
            print(error_message)

def main():
    # 1. Welcome Message
    print("\n--- Welcome to the Expense Classifier and Budget Checker! ---")
    print("This tool will help you understand your spending and manage your monthly budget.\n")

    # 2. User Input with Validation
    monthly_income = get_valid_input(
        "Enter your total monthly income (e.g., 3000): ",
        int,
        lambda x: x > 0,
        "Monthly income must be a positive number."
    )

    monthly_budget_percentage = get_valid_input(
        "Enter the percentage of your income you want to allocate as your monthly budget (1-100, e.g., 60): ",
        int,
        lambda x: 1 <= x <= 100,
        "Monthly budget percentage must be between 1 and 100."
    )

    expense_amount = get_valid_input(
        "Enter the current expense amount you are considering (e.g., 250.50): ",
        float,
        lambda x: x >= 0,
        "Expense amount cannot be negative."
    )

    expense_category = input("Enter the expense category (Groceries, Utilities, Entertainment, Transport, Rent, Savings, Other): ").strip().capitalize()
    valid_categories = ['Groceries', 'Utilities', 'Entertainment', 'Transport', 'Rent', 'Savings', 'Other']
    while expense_category not in valid_categories:
        print("Invalid category. Please choose from: Groceries, Utilities, Entertainment, Transport, Rent, Savings, Other.")
        expense_category = input("Enter the expense category: ").strip().capitalize()

    # 4. Calculations
    total_monthly_budget = monthly_income * (monthly_budget_percentage / 100)
    remaining_budget = total_monthly_budget - expense_amount

    print("\n--- Expense Analysis --- ")

    # 5. Expense Classification and Feedback
    if expense_category == 'Groceries':
        print("Groceries are an essential expense.")
        if expense_amount > 0.20 * total_monthly_budget:
            print("Consider looking for deals or planning your meals to reduce grocery costs.")
    elif expense_category == 'Utilities':
        print("Utilities are necessary for daily living.")
        if expense_amount > 0.15 * total_monthly_budget:
            print("Warning: This utility expense is a bit high. Look for ways to conserve energy or water.")
    elif expense_category == 'Entertainment':
        print("Entertainment is a discretionary expense.")
        if expense_amount > 0.10 * total_monthly_budget:
            print("Caution: This entertainment expense takes a noticeable chunk of your budget.")
        if expense_amount > 0.20 * total_monthly_budget:
            print("Perhaps consider alternative free or low-cost activities for entertainment.")
    elif expense_category == 'Transport':
        print("Transport is essential for most.")
        if expense_amount > 0.15 * total_monthly_budget:
            print("Warning: Your transport expense is quite high. Can you find more economical ways to travel?")
    elif expense_category == 'Rent':
        print("Rent is often a significant fixed cost.")
        if expense_amount > 0.30 * monthly_income:
            print("Strong Warning: Your rent is more than 30% of your monthly income. This might indicate affordability issues.")
    elif expense_category == 'Savings':
        print("Excellent! Investing in your savings is a great financial habit. Keep it up!")
    elif expense_category == 'Other':
        print("This is an 'Other' category expense.")
        if expense_amount > 0.10 * total_monthly_budget:
            print("Suggestion: Review this 'Other' expense carefully. Is it a recurring cost that could be reclassified or reduced?")

    # 6. Overall Budget Check
    print("\n--- Overall Budget Impact ---")
    if expense_amount <= total_monthly_budget:
        if expense_amount < 0.05 * total_monthly_budget:
            print("This expense is a very small part of your budget. Good job!")
        elif 0.05 * total_monthly_budget <= expense_amount <= 0.20 * total_monthly_budget:
            print("This expense fits comfortably within your budget.")
        else: # expense_amount > 0.20 * total_monthly_budget
            print("This expense takes a significant portion of your budget. Consider if it's truly necessary.")
    else:
        print("Warning! This expense exceeds your monthly budget. You need to adjust your spending or reconsider this purchase.")

    # 7. Summary
    print("\n--- Summary --- ")
    print(f"Monthly Income: ${monthly_income:,.2f}")
    print(f"Allocated Monthly Budget: ${total_monthly_budget:,.2f}")
    print(f"Current Expense Amount: ${expense_amount:,.2f}")
    print(f"Expense Category: {expense_category}")
    print(f"Remaining Budget (after this expense): ${remaining_budget:,.2f}")

if __name__ == '__main__':
    main()
````
