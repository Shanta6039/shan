## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
My Budget Tracker Prompt:
Write a Python program that first asks the user for their monthly income. Then, it should allow the user to input up to three separate expenses. After collecting the income and expenses, the program must calculate the remaining balance (income minus total expenses). Finally, it should print a summary of the income, total expenses, and the leftover amount. If the leftover amount is negative, the program should display a clear warning message indicating that the user is over budget.
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# Get monthly income
income = float(input("What is your monthly income? $"))

# Collect up to 3 expenses
expenses = []
print("\nEnter up to 3 expenses (type 'done' to stop early):")

for i in range(3):
    expense_input = input(f"Expense {i + 1}: $")
    if expense_input.lower() == "done":
        break
    try:
        expense_value = float(expense_input)
        expenses.append(expense_value)
    except ValueError:
        print("Invalid input. Please enter a number or 'done'. Skipping this expense.")

# Calculate totals
total_expenses = sum(expenses)
leftover = income - total_expenses

# Print summary
print("\n" + "="*40)
print("BUDGET SUMMARY")
print("="*40)
print(f"Monthly income:   ${income:.2f}")
print(f"Total expenses:   ${total_expenses:.2f}")
print(f"Leftover:         ${leftover:.2f}")

# Warn if negative
if leftover < 0:
    print("\n⚠️ Warning: You are over budget this month!")
else:
    print("\n✅ You're within budget.")
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 A detailed budget summary with advice


```python
enhancement_choice = "I chose to add: Input validation (no negative numbers for income or expenses)"
enhancement_prompt = """
Enhance the existing budget tracker. Add input validation for monthly income and all expenses to ensure that only non-negative numbers are accepted. If a negative number is entered, prompt the user again until a valid non-negative number is provided.
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER

# Function to get validated non-negative float input
def get_positive_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Input cannot be negative. Please enter a non-negative number.")
            else:
                return value
        except ValueError:
            print("Invalid input. Please enter a number.")

# Get monthly income with validation
income = get_positive_float_input("What is your monthly income? $")

# Collect up to 3 expenses with validation
expenses = []
print("\nEnter up to 3 expenses (type 'done' to stop early):")

for i in range(3):
    while True:
        expense_input = input(f"Expense {i + 1}: $")
        if expense_input.lower() == "done":
            break
        try:
            expense_value = float(expense_input)
            if expense_value < 0:
                print("Expense cannot be negative. Please enter a non-negative number.")
            else:
                expenses.append(expense_value)
                break # Exit inner loop if valid input
        except ValueError:
            print("Invalid input. Please enter a number or 'done'.")
    if expense_input.lower() == "done":
        break

# Calculate totals
total_expenses = sum(expenses)
leftover = income - total_expenses

# Print summary
print("\n" + "="*40)
print("BUDGET SUMMARY")
print("="*40)
print(f"Monthly income:   ${income:.2f}")
print(f"Total expenses:   ${total_expenses:.2f}")
print(f"Leftover:         ${leftover:.2f}")

# Warn if negative
if leftover < 0:
    print("\n⚠️ Warning: You are over budget this month!")
else:
    print("\n✅ You're within budget.")
```
