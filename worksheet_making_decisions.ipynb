Here's the Python code to classify a student's academic standing based on their GPA, using the rules we defined earlier:

````python
def classify_academic_standing(gpa):
    """Classifies a student's academic standing based on GPA."""
    if gpa >= 3.5:
        return "Dean's List"
    elif gpa >= 3.0: # Implies gpa < 3.5
        return "Good Standing"
    elif gpa >= 2.0: # Implies gpa < 3.0
        return "Probation"
    else: # Implies gpa < 2.0
        return "Failing"

# Test cases based on your predictions
students = {
    "Student A": 3.7,
    "Student B": 2.9,
    "Student C": 3.0
}

print("Academic Standing Predictions:")
for name, gpa in students.items():
    status = classify_academic_standing(gpa)
    print(f"{name} (GPA: {gpa}): {status}")
````

### Activity 2: Design a study recommendation system

A learning app wants to give students personalised study advice. You design the decision logic.

Pick 2 to 3 factors that matter most to you, for example:

- Hours studied today
- Stress level
- Days until the next exam
- Hours of sleep last night

> **Your design** (double-click to edit):

```
My factors:
- Hours studied today
- Stress level (1-10, 10 being highest)
- Days until the next exam

My rules:

IF (Days until exam is 1 OR Hours studied today is 0) THEN recommend "High-intensity study session! Prioritize exam material."
ELIF (Stress level > 7 AND Hours studied today < 2) THEN recommend "Focus on stress reduction. Light review today, get good sleep."
ELIF (Hours studied today >= 4 AND Stress level < 5) THEN recommend "You're doing great! Consider a short break, then review difficult topics."
ELSE recommend "Continue with your regular study plan."
```

````python
# Your study recommendation system

def get_study_recommendation(hours_studied, stress_level, days_until_exam):
    if days_until_exam == 1 or hours_studied == 0:
        return "High-intensity study session! Prioritize exam material."
    elif stress_level > 7 and hours_studied < 2:
        return "Focus on stress reduction. Light review today, get good sleep."
    elif hours_studied >= 4 and stress_level < 5:
        return "You're doing great! Consider a short break, then review difficult topics."
    else:
        return "Continue with your regular study plan."

# Try these test cases once it works:
# - studied 6 hours, stress level 8, exam in 5 days
# - studied 1 hour, stress level 3, exam in 7 days
# - studied 0 hours, stress level 5, exam tomorrow

print("Test Cases for Study Recommendation System:")

# Test Case 1
recommendation1 = get_study_recommendation(6, 8, 5)
print(f"Studied 6 hours, stress level 8, exam in 5 days: {recommendation1}")

# Test Case 2
recommendation2 = get_study_recommendation(1, 3, 7)
print(f"Studied 1 hour, stress level 3, exam in 7 days: {recommendation2}")

# Test Case 3 (exam tomorrow, 0 hours studied)
recommendation3 = get_study_recommendation(0, 5, 1)
print(f"Studied 0 hours, stress level 5, exam tomorrow: {recommendation3}")

# Additional test case to hit 'Good Standing' or 'Regular Study'
recommendation4 = get_study_recommendation(4, 3, 10)
print(f"Studied 4 hours, stress level 3, exam in 10 days: {recommendation4}")
````

The problem is: The code incorrectly classifies a GPA of 3.2 as 'Probation' when it should be 'Good Standing'.

The fix: The order of conditions in an `if/elif/else` statement matters. More specific or higher-threshold conditions should be checked first.

### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: GPA (Grade Point Average)
Factor 2: Standardized Test Score (e.g., SAT/ACT converted to a single score)
Factor 3: Essay Score (on a scale of 1-10)
Factor 4: Extracurricular Activities (e.g., 'excellent', 'good', 'average', 'none')

Decision logic:
- Eligible if: GPA >= 3.5 AND Test Score >= 1300 AND Essay Score >= 8
- OR (Alternative Path 1): GPA >= 3.0 AND Test Score >= 1200 AND Extracurricular Activities in ('excellent', 'good')
- OR (Alternative Path 2 - Special Consideration): GPA >= 2.5 AND Essay Score >= 9 AND Extracurricular Activities == 'excellent' (for strong applicants with slightly lower GPA)

Thresholds:
- GPA: 3.5 (high), 3.0 (medium), 2.5 (special consideration)
- Test Score: 1300 (high), 1200 (medium)
- Essay Score: 8 (high), 9 (very high)
- Extracurricular Activities: 'excellent', 'good'

Test Cases:
1. Applicant A: GPA=3.7, Test=1350, Essay=8, EC='good' -> Expected: Eligible (Path 1)
2. Applicant B: GPA=3.2, Test=1250, Essay=7, EC='excellent' -> Expected: Eligible (Alternative Path 1)
3. Applicant C: GPA=2.8, Test=1100, Essay=9, EC='excellent' -> Expected: Eligible (Alternative Path 2)
4. Applicant D: GPA=3.0, Test=1150, Essay=7, EC='average' -> Expected: Not Eligible
```

````python
# Your admissions decision system

def check_admissions_eligibility(gpa, test_score, essay_score, extracurriculars):
    # Path 1: High academic achievement
    path1_eligible = (gpa >= 3.5 and test_score >= 1300 and essay_score >= 8)

    # Alternative Path 1: Good academic achievement with strong extracurriculars
    path2_eligible = (gpa >= 3.0 and test_score >= 1200 and extracurriculars in ('excellent', 'good'))

    # Alternative Path 2: Special consideration for lower GPA but outstanding essay and ECs
    path3_eligible = (gpa >= 2.5 and essay_score >= 9 and extracurriculars == 'excellent')

    return path1_eligible or path2_eligible or path3_eligible


print("=== University Admissions Eligibility ===")

# Test Cases
applicants = [
    {"name": "Applicant A", "gpa": 3.7, "test": 1350, "essay": 8, "ec": "good"},
    {"name": "Applicant B", "gpa": 3.2, "test": 1250, "essay": 7, "ec": "excellent"},
    {"name": "Applicant C", "gpa": 2.8, "test": 1100, "essay": 9, "ec": "excellent"},
    {"name": "Applicant D", "gpa": 3.0, "test": 1150, "essay": 7, "ec": "average"}
]

for applicant in applicants:
    eligible = check_admissions_eligibility(
        applicant["gpa"],
        applicant["test"],
        applicant["essay"],
        applicant["ec"]
    )
    status = "Eligible" if eligible else "Not Eligible"
    print(f"{applicant['name']} (GPA: {applicant['gpa']}, Test: {applicant['test']}, Essay: {applicant['essay']}, EC: {applicant['ec']}): {status}")
````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
